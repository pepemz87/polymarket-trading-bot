# 📤 Subir Bot a GitHub desde Google Colab

Este notebook te ayuda a subir el código del bot a GitHub **sin usar tu computadora local**.

## 📋 Requisitos Previos
1. Cuenta de GitHub (gratis) - [Crear aquí](https://github.com/signup)
2. Crear un repositorio vacío en GitHub:
   - Ve a https://github.com/new
   - Nombre: `polymarket-trading-bot`
   - Público o Privado (tu elección)
   - **NO marques** "Add a README file"
   - Click "Create repository"
3. Crear un Personal Access Token:
   - Ve a https://github.com/settings/tokens
   - Click "Generate new token" → "Generate new token (classic)"
   - Nombre: `colab-upload`
   - Marca: `repo` (todos los permisos de repositorio)
   - Click "Generate token"
   - **COPIA EL TOKEN** (solo se muestra una vez)

## 🚀 Pasos
Ejecuta cada celda en orden.

## Paso 1: Subir el Código a Colab

**Opción A: Subir ZIP**
1. En tu PC, comprime la carpeta `molten-magnetar` en un ZIP
2. Ejecuta la celda siguiente
3. Click en "Choose Files" y selecciona el ZIP

````python
from google.colab import files
import zipfile
import os

print("📤 Sube el archivo ZIP de tu bot...")
uploaded = files.upload()

# Obtener nombre del archivo
zip_name = list(uploaded.keys())[0]

# Descomprimir
print(f"\n📦 Descomprimiendo {zip_name}...")
with zipfile.ZipFile(zip_name, 'r') as zip_ref:
    zip_ref.extractall('/content/')

# Buscar la carpeta del proyecto
!ls -la /content/

print("\n✅ Código subido y descomprimido")
````

## Paso 2: Ir a la Carpeta del Proyecto

````python
# Cambiar al directorio del proyecto
# Ajusta el nombre si tu carpeta se llama diferente
%cd /content/molten-magnetar

# Verificar contenido
!ls -la

print("\n✅ En el directorio del proyecto")
````

## Paso 3: Configurar Git

````python
from getpass import getpass

# Solicitar información
print("🔧 Configuración de Git")
print("="*50)

git_name = input("Tu nombre (para commits): ")
git_email = input("Tu email de GitHub: ")
github_username = input("Tu usuario de GitHub: ")
github_token = getpass("Tu Personal Access Token de GitHub: ")

# Configurar git
!git config --global user.name "{git_name}"
!git config --global user.email "{git_email}"

print("\n✅ Git configurado")
````

## Paso 4: Inicializar Repositorio Git

````python
# Inicializar git
!git init

# Agregar todos los archivos
!git add .

# Hacer commit
!git commit -m "Initial commit - Polymarket trading bot"

# Crear rama main
!git branch -M main

print("\n✅ Repositorio inicializado")
````

## Paso 5: Conectar con GitHub y Subir

````python
# Crear URL con token
repo_url = f"https://{github_token}@github.com/{github_username}/polymarket-trading-bot.git"

# Agregar remote
!git remote add origin {repo_url}

# Push a GitHub
print("📤 Subiendo a GitHub...")
!git push -u origin main

print("\n✅ ¡Código subido a GitHub!")
print(f"\n🔗 Tu repositorio: https://github.com/{github_username}/polymarket-trading-bot")
````
